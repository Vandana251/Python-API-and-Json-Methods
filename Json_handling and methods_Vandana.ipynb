```python
# (1.) GET - Display User Details 

# Retrieve user information from the API. 

import requests 

users_url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(users_url)
users = response.json()
print(users)
print("-----------------------------------------------------------------------")

# To get entire data in structured format we use .text 

print(response.text)
```

```python
# Display User Details          ( TASK - 1 )

for user in users:
    print("ID:", user["id"])
    print("Name:", user["name"])
    print("Username:", user["username"])
    print("Email:", user["email"])
    print("Phone:", user["phone"])
    print("-" * 40) 
```

```python
# TASK - 2 ( Display All Posts )

posts_url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(posts_url)

posts = response.json()

for post in posts:
    print("Post ID:", post["id"])
    print("User ID:", post["userId"])
    print("Title:", post["title"])
    print("Body:", post["body"])
    print("-" * 50)
```

```python
# POST — Create a New Post 

# POST is used to send data to the server and create a new resource.

import requests

URL = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "Learning REST API",
    "body": "I am learning API operations using Python.",
    "userId": 1
}

response = requests.post(URL, json=data)

print("Status Code:", response.status_code)
print("Response:", response.json())
```

```python
# PUT — Complete Update = PUT is used to completely replace/update an existing resource.

import requests

URL = "https://jsonplaceholder.typicode.com/posts/1"

data = {
    "id": 1,
    "title": "Updated Title",
    "body": "Updated post content",
    "userId": 1
}

response = requests.put(URL, json=data)

print("Status Code:", response.status_code)
print("Response:", response.json())
```

```python
# 5. PATCH — Partial Update = PATCH is used to update only specific fields of an existing resource.

import requests

URL = "https://jsonplaceholder.typicode.com/posts/1"

data = {
    "title": "Only Title Updated"
}

response = requests.patch(URL, json=data)

print("Status Code:", response.status_code)
print("Response:", response.json())

# PUT -> Replace/update complete resource

# PATCH -> Update only selected fields
```

```python
6. # DELETE — Delete a Post = DELETE is used to remove an existing resource

import requests

URL = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(URL)

print("Status Code:", response.status_code)
```

Summary:- 

I worked with the JSONPlaceholder REST API to understand how client applications communicate with RESTful services. I retrieved and displayed user details and posts using GET. I also practiced creating, updating, partially updating, and deleting resources using POST, PUT, PATCH, and DELETE. This helped me understand the basic CRUD operations and how different HTTP methods are used in REST APIs.
